```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv(filename)

# Display column names
print(df.columns)

# Line chart
plt.figure(figsize=(10, 5))

plt.plot(df.index, df.iloc[:, 0], marker='o')

plt.title("Line Chart")
plt.xlabel("Student / Record")
plt.ylabel(df.columns[0])

plt.grid(True)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv(filename)

# Histogram
plt.figure(figsize=(8, 5))

plt.hist(df.iloc[:, 0], bins=10, edgecolor='black')

plt.title("Histogram")
plt.xlabel(df.columns[0])
plt.ylabel("Frequency")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv(filename)

# Bar chart
plt.figure(figsize=(10, 5))

# Display first 10 records
plt.bar(df.index[:10], df.iloc[:10, 0])

plt.title("Bar Chart")
plt.xlabel("Student / Record")
plt.ylabel(df.columns[0])

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv(filename)

# Find numerical columns
numeric_columns = df.select_dtypes(include='number').columns

print("Numerical columns:")
print(numeric_columns)

# Select first two numerical columns
x = numeric_columns[0]
y = numeric_columns[1]

# Scatter plot
plt.figure(figsize=(8, 5))

plt.scatter(df[x], df[y])

plt.title("Positive Relation")
plt.xlabel(x)
plt.ylabel(y)

plt.grid(True)
plt.show()

# Calculate correlation
correlation = df[x].corr(df[y])

print("Correlation =", correlation)

if correlation > 0:
    print("The variables have a Positive Relation.")
else:
    print("The selected variables do not have a positive relation.")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv(filename)

# Find numerical columns
numeric_columns = df.select_dtypes(include='number').columns

print("Numerical columns:")
print(numeric_columns)

# Select two numerical columns
x = numeric_columns[0]
y = numeric_columns[1]

# Scatter plot
plt.figure(figsize=(8, 5))

plt.scatter(df[x], df[y])

plt.title("Negative Relation")
plt.xlabel(x)
plt.ylabel(y)

plt.grid(True)
plt.show()

# Calculate correlation
correlation = df[x].corr(df[y])

print("Correlation =", correlation)

if correlation < 0:
    print("The variables have a Negative Relation.")
else:
    print("The selected variables do not have a negative relation.")
```
